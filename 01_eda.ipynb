```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('synthetic_rentals.csv')

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(
    data=df,
    x="price",
    bins=40,
    log_scale=True,
    color="#4C78A8",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Distribution of Rental Prices (Log Scale)")
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Count")

sns.scatterplot(
    data=df,
    x="sqft",
    y="price",
    alpha=0.65,
    color="#F58518",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Price vs. Square Footage")
axes[0, 1].set_xlabel("Sqft")
axes[0, 1].set_ylabel("Price")

sns.boxplot(
    data=df,
    x="neighborhood",
    y="price",
    palette="Set2",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Price by Neighborhood")
axes[1, 0].set_xlabel("Neighborhood")
axes[1, 0].set_ylabel("Price")
axes[1, 0].tick_params(axis="x", rotation=30)

missing = df.isna().astype(int)
sns.heatmap(
    missing,
    cmap="YlOrRd",
    cbar=True,
    linewidths=0.1,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Missing Value Heatmap")
axes[1, 1].set_xlabel("Columns")
axes[1, 1].set_ylabel("Rows")

plt.tight_layout()
plt.show()
```
